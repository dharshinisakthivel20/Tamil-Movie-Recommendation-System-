```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Tamil_movies_dataset.csv')

# Print all column names
print(df.columns)

# Display the first 5 rows
df.head()
```

```python
# 1. Select only the columns we need for recommendation
movies_df = df[['MovieName', 'Genre', 'Director', 'Actor']]

# 2. Drop any empty rows if present
movies_df.dropna(inplace=True)

# 3. Combine Genre, Director, and Actor into a single column called 'tags'
movies_df['tags'] = movies_df['Genre'] + " " + movies_df['Director'] + " " + movies_df['Actor']

# 4. Create a clean final table
final_df = movies_df[['MovieName', 'tags']]

# Display the first 5 rows of our clean table
final_df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Initialize the TF-IDF Vectorizer
tfidf = TfidfVectorizer(stop_words='english')

# 2. Convert text tags into mathematical vectors (numbers)
tfidf_matrix = tfidf.fit_transform(final_df['tags'])

# 3. Compute the Cosine Similarity Score matrix between all movies
similarity = cosine_similarity(tfidf_matrix)

# Print the final matrix shape to verify
print("Similarity Matrix shape:", similarity.shape)
```

```python
# Create the recommendation engine function
def recommend_movies(movie_title):
    try:
        # 1. Find the index of the given movie name
        movie_idx = final_df[final_df['MovieName'] == movie_title].index[0]

        # 2. Get similarity scores of this movie with all other movies
        similarity_scores = list(enumerate(similarity[movie_idx]))

        # 3. Sort the movies based on highest similarity score, leaving out the same movie
        sorted_movies = sorted(similarity_scores, key=lambda x: x[1], reverse=True)[1:6]

        print(f"If you like '{movie_title}', you should also watch:\n")
        # 4. Print the top 5 movie names
        for item in sorted_movies:
            print(final_df.iloc[item[0]]['MovieName'])

    except IndexError:
        print("Sorry! This movie is not in our dataset. Please try another Tamil movie name.")

# TEST THE MODEL (You can change 'Mankatha' to any movie from your dataset table like '7 Aum Arivu')
recommend_movies('Mankatha')
```

```python
# Save the cleaned final dataframe as a CSV file to download
final_df.to_csv('final_movies.csv', index=False)
print("File saved successfully! Check the left-side folder panel.")
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

st.set_page_config(page_title="Tamil Movie Recommender", page_icon="🎬", layout="centered")
st.title("🎬 Tamil Movie Recommendation Engine")
st.markdown("Developed by **DHARSHINI S** for Placement Showcase")
st.write("Select a movie from the dropdown to get AI-powered similar recommendations.")

df = pd.read_csv('Tamil_movies_dataset.csv')
movies_df = df[['MovieName', 'Genre', 'Director', 'Actor']].dropna()
movies_df['tags'] = movies_df['Genre'] + " " + movies_df['Director'] + " " + movies_df['Actor']
final_df = movies_df[['MovieName', 'tags']].reset_index(drop=True)

tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(final_df['tags'].fillna(''))
similarity = cosine_similarity(tfidf_matrix)

movie_list = final_df['MovieName'].values
selected_movie = st.selectbox("Choose a Tamil Movie:", movie_list)

if st.button('Show Recommendations'):
    try:
        movie_idx = final_df[final_df['MovieName'] == selected_movie].index[0]
        similarity_scores = list(enumerate(similarity[movie_idx]))
        sorted_movies = sorted(similarity_scores, key=lambda x: x[1], reverse=True)[1:6]

        st.success(f"Top 5 Recommendations if you like '{selected_movie}':")
        for item in sorted_movies:
            st.write(f"🍿 {final_df.iloc[item[0]]['MovieName']}")
    except Exception as e:
        st.error("Something went wrong! Please try another movie title.")

```
